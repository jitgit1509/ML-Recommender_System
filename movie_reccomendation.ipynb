```python
import warnings
warnings.filterwarnings('ignore')
```

```python
import numpy as np
import pandas as pd
```

```python
user_data=pd.read_csv('user_data.csv')
user_data.head()
```

```python
movies=pd.read_csv('Movie_Id_Titles.csv')
movies.head()
```

```python
df=pd.merge(user_data,movies, on='item_id')
```

```python
df.head()
```

```python
movie_matrix=df.pivot_table(index='user_id', columns='title', values='rating')
movie_matrix
```

```python
movie_matrix.to_csv('movie_matrix.csv')
```

```python
startwars_ratings = movie_matrix['Star Wars (1977)']
startwars_ratings
```

```python
similar_to_starwars = movie_matrix.corrwith(startwars_ratings).sort_values(ascending=False)
similar_to_starwars
```

```python
corr_starwars=pd.DataFrame(similar_to_starwars, columns=['Correlation'])
```

```python
corr_starwars.dropna(inplace=True)
```

```python
corr_starwars
```

```python
ratings = pd.DataFrame(df.groupby('title')['rating'].count())
```

```python
ratings
```

```python
corr_starwars=corr_starwars.join(ratings['rating'])
corr_starwars
```

```python
corr_starwars[corr_starwars['rating']>100]
```

```python
liarliar_ratings = movie_matrix['Liar Liar (1997)']
similar_to_liarliar = movie_matrix.corrwith(liarliar_ratings).sort_values(ascending=False)
corr_liarliar=pd.DataFrame(similar_to_liarliar, columns=['Correlation'])
corr_liarliar.dropna(inplace=True)
corr_liarliar=corr_liarliar.join(ratings['rating'])
corr_liarliar[corr_liarliar['rating']>100]
```

```python
RB_ratings = movie_matrix['Raging Bull (1980)']
similar_to_RB = movie_matrix.corrwith(RB_ratings).sort_values(ascending=False)
corr_RB=pd.DataFrame(similar_to_RB, columns=['Correlation'])
corr_RB.dropna(inplace=True)
corr_RB=corr_RB.join(ratings['rating'])
corr_RB[corr_RB['rating']>100]
```
